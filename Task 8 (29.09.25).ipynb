```python
# 2.Create a Student class with a constructor that takes name, age, and grade. Display student information using a method.
class student:
    def __init__(self,name,age,grade):
        self.name = name
        self.age = age
        self.grade = grade

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Grade: {self.grade}")

s1 = student("Janani",24,"A Grade")
s2 = student("Mubarak",24,"B Grade")
s3 = student("Lokesh",25,"C Grade")

s1.display_info()
s2.display_info()
s3.display_info()
        





    
```
