```python
# 2.Create a Student class with a constructor that takes name, age, and grade. Display student information using a method.
class student:
    def __init__(self,name,age,grade):
        self.name = name
        self.age = age
        self.grade = grade

s1 = student('Janani',24,'A')
print(f"Name: {s1.name}")
print(f"Age: {s1.age}")
print(f"Grade: {s1.grade}")

    
```
