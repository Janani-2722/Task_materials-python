```python
# 2.Bank Account Management System
# Create a class BankAccount with deposit, withdraw, and balance check methods.
# Maintain a transaction history using a list.
# Overload the + operator to merge balances of two accounts.
class BankAccount:
    def __init__(self, Account_holder, initial_balance=0):
        self.Account_holder =  Account_holder
        self.balance = initial_balance
        self.history = []  

    def deposit(self, amount):
        self.balance += amount
        self.history.append(f"Deposited {amount}. Current balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
            self.history.append(f"Failed withdrawal {amount}")
        else:
            self.balance -= amount
            self.history.append(f"Withdraw {amount}. Current Balance: {self.balance}")

    def check_balance(self):
        print(f"Balance for {self.Account_holder}: {self.balance}")
        return self.balance

    def transaction_history(self):
        print(f"Transaction history for {self.Account_holder}:")
        for transaction in self.history:
            print(transaction)

    def __add__(self, other):
        new_Account_holder = f"{self.Account_holder} & {other.Account_holder}"
        new_balance = self.balance + other.balance
        new_account = BankAccount(new_Account_holder, new_balance)
        
        new_account.history = self.history + other.history + [f"Merged the accounts. Here, New balance: {new_balance}"]
        return new_account


account1 = BankAccount("Janani", 1000)
account2 = BankAccount("Keerthana", 500)

account1.deposit(500)
account1.withdraw(200)

account2.deposit(500)
account2.withdraw(500)

account1.check_balance()
account2.check_balance()

merged_account = account1 + account2
merged_account.check_balance()
merged_account.transaction_history()

 

```
