```python
# 2.Base class: Employee with method calculate_salary().
# Derived classes:
# FullTimeEmployee – uses a fixed monthly salary.
# PartTimeEmployee – calculates based on hours worked.
class Employee:
    def calculate_salary(self):
        pass

class FullTimeEmployee(Employee):
    def __init__(self, salary):
        self._salary = salary #protected variable

    def calculate_salary(self):
        return self._salary

class PartTimeEmployee(Employee):
    def __init__(self, rate, hours):
        self._rate = rate
        self._hours = hours

    def calculate_salary(self):
        return self._rate * self._hours

f = FullTimeEmployee(30000)
p = PartTimeEmployee(200, 50)

print("Full Time Salary:", f.calculate_salary())
print("Part Time Salary:", p.calculate_salary())

            
#polymorphism- same method name different classes
```
