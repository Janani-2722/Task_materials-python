```python
# 2.Write a script to read a file and print the number of lines in it.
count = 0
with open("file","a+") as f: #'a+' - create file,append,read
    f.write("This is the python file.\n")
    f.write("This will act as library.\n")
    f.write("Here, we will run the lines.")
    f.seek(0)
    content = f.readlines()
    # print(content)
    for i in content:
        count += 1
    print(f" No of lines in the file: {count} lines")
    
```
