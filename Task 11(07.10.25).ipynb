```python
# 2.Create a parameterized decorator repeat(n) that executes a function n times.
def repeat(n):
    print("Starting to execution the function")
    def inner(func):
        print(f"Function executes upto {n} times.") 
        def wrapper():
            print("Execution starts!")
            for i in range(1,n +1):    
                print(f"Execution {i}:")
                # i+=1
                func()
            print("Execution stops!")
        return wrapper
    return inner

@repeat(n = 5)
def my_func():
    print("Function Executing")

my_func()
```
