```python
# 1.Create a 2D NumPy array (10×5) with random integers (1–100).
import numpy as np

arr = np.random.randint(1, 101, (10, 5))
print(arr)
```

```python
# 2. Replace all values < 50 with 0 and ≥ 50 with 1.
binary_arr = np.where(arr < 50, 0, 1)
print(binary_arr)
 
```

```python
# 3. Compute mean, standard deviation, variance along rows and columns.
row_mean = arr.mean(axis=1)
col_mean = arr.mean(axis=0)

row_std = arr.std(axis=1)
col_std = arr.std(axis=0)

row_var = arr.var(axis=1)
col_var = arr.var(axis=0)
```

```python
# 4. Generate two arrays A(3×3) and B(3×3) and perform matrix multiplication.
A = np.random.randint(1, 10, (3,3))
B = np.random.randint(1, 10, (3,3))

result = A @ B
print(result)
```

```python
# 5. Find the eigenvalues and eigenvectors of a 3×3 matrix.
vals, vecs = np.linalg.eig(A)
print("Eigenvalues:", vals)
print("Eigenvectors:\n", vecs)
```

```python
# 6. Create a 1D array of 100 random values and normalize it (0–1 scale).
arr1d = np.random.rand(100)
norm = (arr1d - arr1d.min()) / (arr1d.max() - arr1d.min())
```

```python
# 7. Stack two arrays vertically and horizontally.
x = np.array([1,2,3])
y = np.array([4,5,6])

vstack = np.vstack((x,y))
hstack = np.hstack((x,y))
```

```python
# 8. Create an array of shape (5,5) and extract the diagonal, upper triangle and lower triangle.
m = np.random.randint(1, 20, (5,5))

diag = np.diag(m)
upper = np.triu(m)
lower = np.tril(m)
```

```python
# 9. Generate a 4×4 array and compute its determinant and inverse.
mat = np.random.randint(1, 10, (4,4))

det = np.linalg.det(mat)
inv = np.linalg.inv(mat)
```

```python
# 10. Write a NumPy program to perform element-wise comparison between two equal-sized arrays.
a = np.array([1,2,3])
b = np.array([3,2,1])

print(a == b)
print(a > b)
print(a < b)
```

```python
# 11. Create a 1000-element array and compute: 25th percentile, 50th percentile, 75th percentile
big = np.random.rand(1000)

p25 = np.percentile(big, 25)
p50 = np.percentile(big, 50)
p75 = np.percentile(big, 75)
```
