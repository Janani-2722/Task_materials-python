```python
# 2.Create a function process_file() that contains nested functions for:
# Counting lines
# Counting words
# Counting characters
# Return all three as a dictionary.

def process_file(file):
    with open("new_file.txt", "w") as f:
        f.write("My name is Janani.\nI am learning AI/ML Course.")
    with open(file, "r") as file:
        content = file.read()
        
    def count_lines():
        lines = content.split('\n')
        return len(lines)
    def count_words():
        words = content.split()
        return len(words)
    def count_char():
        return len(content)
    return {
        'lines': count_lines(),
        'words': count_words(),
        'characters': count_char()
    }
    
print(process_file("new_file.txt"))

```
