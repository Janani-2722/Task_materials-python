```python
# 2.Create a class Employee with a property salary. Use @property for getting and setting to ensure the salary is a positive number.
class Employee:
    def __init__(self,salary = 0):
        self.__salary = salary
        
    def emp_salary(self):
        return self.__salary

    @property
    def salary(self):
        print("Getting salary...")
        return self.__salary

    @salary.setter
    def salary(self, value):
        print("Setting Salary...")
        if value <= 0:
            raise ValueError("Salary should be Positive neither Negative nor Zero")
        self.__salary = value

person = Employee()
person.salary = 30000
print(person.emp_salary())
```
